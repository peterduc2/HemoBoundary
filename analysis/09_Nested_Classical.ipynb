# Nested classical comparisons

Runs locally in Colab using public research code. Data and results remain on Google Drive.


```python
from pathlib import Path
import subprocess
import sys
from google.colab import drive

drive.mount('/content/gdrive', force_remount=False)
repo = Path('/content/HemoBoundary')
if not (repo / 'scripts' / 'audit_structure.py').is_file():
    if repo.exists():
        raise RuntimeError('Please use a clean /content/HemoBoundary location')
    subprocess.run(['git', 'clone', '--depth', '1',
                    'https://github.com/peterduc2/HemoBoundary.git', str(repo)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q',
                '-r', str(repo / 'requirements.txt')], check=True)
subprocess.run([sys.executable, str(repo / 'scripts' / 'audit_structure.py')], check=True)
if not (repo / 'src' / 'recorded_background.py').is_file():
    raise RuntimeError('This checkout does not include the updated recorded-background source')
project = Path('/content/gdrive/MyDrive/HonorProject')
shin = project / 'shin2018_word_generation_state_preparation_v1'
if not (shin / '95_N4_FUNCTIONAL_ABC_LBR_FULLY_NESTED_10SEEDS_V1').is_dir():
    shin = project / 'HemoBoundary' / 'SHIN'
rest = project / 'HemoBoundary' / 'Luhmann2020_Resting'
completed = project / 'HEMOBOUNDARY_ARRAY_THREE_EXPERIMENTS_20261102'
archive = project / 'HEMOBOUNDARY_ARRAY_FOUR_VALIDATIONS_20261025.zip'
cmd = [sys.executable, str(repo / 'scripts' / 'run_nested_classical.py'),
       '--shin-root', str(shin), '--output', str(completed)]
if archive.is_file():
    cmd += ['--original-archive', str(archive)]
subprocess.run(cmd, check=True)
print('Nested inference and participant summaries saved outside the repository:', completed)
```
